```python
while True :
    n = int(input("NHAP SO LUONG SINH VIEN : "))
    if n > 0 :
        break
    print("NHAP LAI : ")

diem = []
for i in range(n):
    while True :
        x = float(input("NHAP GIA TRI : "))
        if x >= 0 and x <= 10:
            diem.append(x)
            break
        else : print("NHAP LAI : ")

tong = 0
for i in diem :
    tong += i
print("GIA TRI TRUNG BINH : ", round(tong/len(diem), 2))

maxx = diem[0]
minn = diem[0]
for i in diem:
    if i > maxx:
        maxx = i
    if i < minn :
        minn = i
print("GIA TRI LON NHAT : ", maxx)
print("GIA TRI NHO NHAT : ", minn)

tren_tb = []
for i in diem :
    if i > (tong/len(diem)) :
        tren_tb.append(i)
print("DIEM TREN TRUNG BINH : ", tren_tb)

for i in diem:
    if i == 10:
        print("CO DIEM 10!")
        break
else: Print("KHONG CO DIEM 10!")
```
