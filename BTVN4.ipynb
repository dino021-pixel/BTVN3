```python
while True:
    n = int(input("Nhap so luong khoan chi: "))
    if n > 0:
        break
    print("Nhap lai!")

ds = []


for i in range(n):
    s = input("Nhap khoan chi: ")
    x = s.split(",")

    ten = x[0].strip()
    tien = int(x[1].strip())
    danh_muc = x[2].strip()

    ds.append((ten, tien, danh_muc))


print("\nDanh sach cac khoan chi:")
for i in ds:
    print(i)


tong = 0
for i in ds:
    tong += i[1]

print("\nTong chi tieu:", tong, "VND")


print("\nThong ke theo danh muc:")

danh_muc = []

for i in ds:
    if i[2] not in danh_muc:
        danh_muc.append(i[2])

for dm in danh_muc:
    dem = 0
    tong_dm = 0

    for i in ds:
        if i[2] == dm:
            dem += 1
            tong_dm += i[1]

    print("\n" + dm + ":")
    print("- So khoan chi:", dem)
    print("- Tong tien:", tong_dm, "VND")


if tong > 5000000:
    print("\nTong chi tieu vuot qua 5.000.000 VND.")


lon_nhat = ds[0]

for i in ds:
    if i[1] > lon_nhat[1]:
        lon_nhat = i

print("\nKhoan chi co so tien lon nhat:")
print(lon_nhat)
```
