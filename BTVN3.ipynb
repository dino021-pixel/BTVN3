```python
a = input("Nhap so thich cua nguoi A: ")
b = input("Nhap so thich cua nguoi B: ")


ds_a = a.split(",")
for i in range(len(ds_a)):
    ds_a[i] = ds_a[i].strip().title()

ds_b = b.split(",")
for i in range(len(ds_b)):
    ds_b[i] = ds_b[i].strip().title()


set_a = set(ds_a)
set_b = set(ds_b)


print("Cac so thich cua Nguoi A:")
print(set_a)

print("\nCac so thich cua Nguoi B:")
print(set_b)


chung = set_a & set_b
print("\nSo thich chung:")
if len(chung) == 0:
    print("Khong co so thich chung.")
else:
    print(chung)


chi_a = set_a - set_b
print("\nSo thich chi Nguoi A co:")
print(chi_a)


tat_ca = set_a | set_b
print("\nTat ca so thich:")
print(tat_ca)


if len(tat_ca) == 0:
    do_tuong_dong = 0
else:
    do_tuong_dong = len(chung) / len(tat_ca) * 100

print("\nDo tuong dong: {:.2f}%".format(do_tuong_dong))
```
