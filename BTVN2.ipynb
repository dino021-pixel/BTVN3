```python
s = input("Nhap cac san pham: ")
kiem_tra = input("Nhap san pham can kiem tra: ")


ds = s.split(",")

for i in range(len(ds)):
    ds[i] = ds[i].strip().title()

print("Danh sach san pham:")
print(ds)


print("\nTong so san pham da mua:", len(ds))


if len(ds) % 2 != 0:
    print("\nSan pham o vi tri giua:", ds[len(ds) // 2])


tap = set(ds)
max_dem = 0

for sp in tap:
    if ds.count(sp) > max_dem:
        max_dem = ds.count(sp)

nhieu_nhat = []
for sp in tap:
    if ds.count(sp) == max_dem:
        nhieu_nhat.append(sp)

nhieu_nhat.sort()

print("\nCac san pham duoc mua nhieu nhat:")
for sp in nhieu_nhat:
    print(sp + ":", max_dem, "lan")


kiem_tra = kiem_tra.strip().title()

if kiem_tra in ds:
    print("\n", kiem_tra, "duoc mua", ds.count(kiem_tra), "lan.")
else:
    print("\n", kiem_tra, "chua duoc mua.")


ds.insert(0, "Bánh Nabati")


if "Sữa" in ds:
    ds.remove("Sữa")

print("\nDanh sach sau khi cap nhat:")
print(ds)
```
