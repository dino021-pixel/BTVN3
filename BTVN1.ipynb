```python
s = input("Nhap chuoi : ")

dao = ""
for i in range(len(s) - 1, -1, -1):
    dao += s[i]
print("Chuoi dao nguoc : ", dao)

sap_xep = "".join(sorted(s))
print("Chuoi sau khi sap xep : ", sap_xep)

if s == dao:
    print("Day la chuoi doi xung")
else:
    print("Day khong phai la chuoi doi xung")

tap_ky_tu = set(s)
max_dem = 0

for c in tap_ky_tu:
    if s.count(c) > max_dem:
        max_dem = s.count(c)

ds = []
for c in tap_ky_tu:
    if s.count(c) == max_dem:
        ds.append(c)

ds.sort()

for c in ds:
    print(c, end=" ")
print()

print("So lan xuat hien:", max_dem)

nguyen_am = {'a', 'e', 'i', 'o', 'u'}
chuoi_thuong = s.lower()

if nguyen_am.issubset(set(chuoi_thuong)):
    print("Chuoi chua day du 5 nguyen am ")
else:
    print("Chuoi khong chua day du 5 nguyen am ")
```
